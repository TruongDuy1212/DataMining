# BÀI 3: KHAI PHÁ LUẬT KẾT HỢP (ASSOCIATION RULES)

**Mục tiêu:** Áp dụng thuật toán Apriori để khai phá các quy tắc ẩn (If-Then) trên 2 bộ dữ liệu thực tế nhằm tìm ra các dấu hiệu đi kèm với một hành vi/kết quả cụ thể:
1. **D3 (NSL-KDD):** Phát hiện các quy luật về giao thức và cờ trạng thái thường đi kèm với một cuộc tấn công mạng (`attack_class_attack`).
2. **D2 (Hotel Bookings):** Nhận diện các điều kiện (loại khách, kênh đặt, đặt cọc) dẫn đến hành vi hủy phòng (`is_canceled_1`).

---
## 3.1. Triển khai thuật toán Apriori trên bộ dữ liệu D3 (NSL-KDD)

```python
import warnings
warnings.filterwarnings('ignore')

# ==========================================
# BÀI 3.1: LUẬT KẾT HỢP APRIORI TRÊN D3 (NSL-KDD)
# ==========================================
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from IPython.display import display, Markdown

display(Markdown("### 🔍 TÌM LUẬT CẢNH BÁO TẤN CÔNG MẠNG (D3)"))

try:
    # 1. Đọc và tiền xử lý dữ liệu
    df3 = pd.read_csv('D3_Cleaned_NSLKDD.csv')
    cat_cols_d3 = ['protocol_type', 'service', 'flag', 'attack_class']
    df_encoded_d3 = pd.get_dummies(df3[cat_cols_d3]).astype(bool)

    # 2. Chạy thuật toán
    frequent_itemsets_d3 = apriori(df_encoded_d3, min_support=0.1, use_colnames=True)
    rules_d3 = association_rules(frequent_itemsets_d3, metric="confidence", min_threshold=0.8)

    # 3. Lọc luật và lấy Top 5
    attack_rules = rules_d3[rules_d3['consequents'] == {'attack_class_attack'}]
    top_attack_rules = attack_rules.sort_values(by='lift', ascending=False).head(5)

    # 4. Hiển thị bảng dạng HTML tuyệt đẹp
    display(Markdown("**=> TOP 5 LUẬT TẤN CÔNG MẠNG NGUY HIỂM NHẤT:**"))
    display(top_attack_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].round(4))

except Exception as e:
    print(f"LỖI: {e}")
```

---
## 3.2. Triển khai thuật toán Apriori trên bộ dữ liệu D2 (Hotel Bookings)

```python
import warnings
warnings.filterwarnings('ignore')

# ==========================================
# BÀI 3.2: LUẬT KẾT HỢP APRIORI TRÊN D2 (HOTEL BOOKINGS)
# ==========================================
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from IPython.display import display, Markdown

display(Markdown("### 🏨 TÌM LUẬT CẢNH BÁO HỦY PHÒNG (D2)"))

try:
    # 1. Đọc và tiền xử lý dữ liệu
    df2 = pd.read_csv('D2_Cleaned_HotelBookings.csv')
    cat_cols_d2 = ['meal', 'market_segment', 'deposit_type', 'customer_type', 'is_canceled']
    df_encoded_d2 = pd.get_dummies(df2[cat_cols_d2].astype(str)).astype(bool)

    # 2. Chạy thuật toán
    frequent_itemsets_d2 = apriori(df_encoded_d2, min_support=0.1, use_colnames=True)
    rules_d2 = association_rules(frequent_itemsets_d2, metric="confidence", min_threshold=0.7)

    # 3. Lọc luật Hủy phòng
    cancel_rules = rules_d2[rules_d2['consequents'] == {'is_canceled_1'}]
    if cancel_rules.empty:
        cancel_rules = rules_d2[rules_d2['consequents'] == {'is_canceled_1.0'}]

    top_cancel_rules = cancel_rules.sort_values(by='lift', ascending=False).head(5)

    # 4. Hiển thị bảng dạng HTML tuyệt đẹp
    display(Markdown("**=> TOP 5 QUY LUẬT DẪN ĐẾN HÀNH VI HỦY PHÒNG:**"))
    display(top_cancel_rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].round(4))

except Exception as e:
    print(f"LỖI: {e}")
```
